```python
# 1. Criação das variáveis com os dados do produto
preco_original = 100.0  # Preço inicial do produto em reais
percentual_desconto = 15  # Porcentagem de desconto (15%)

# 2. Cálculo do valor do desconto
# Dividimos o percentual por 100 para converter a porcentagem em decimal (ex: 15 / 100 = 0.15)
valor_desconto = preco_original * (percentual_desconto / 100)

# 3. Cálculo do preço final
preco_final = preco_original - valor_desconto

# 4. Exibição dos resultados na tela
print("Preço original: R$", preco_original)
print("Desconto aplicado:", percentual_desconto, "%")
print("Valor do desconto: R$", valor_desconto)
print("Preço final com desconto: R$", preco_final)
```

```python
# 1. Entrada de dados fornecida pelo vendedor
nome_vendedor = input("Digite o seu nome: ")
nome_produto = input("Digite o nome do produto: ")

# Convertemos para float para aceitar números com casas decimais (ex: 150.50 e 10.5)
preco_original = float(input("Digite o preço original do produto (R$): "))
percentual_desconto = float(input("Digite a porcentagem de desconto (%): "))

# 2. Cálculos matemáticos
valor_desconto = preco_original * (percentual_desconto / 100)
preco_final = preco_original - valor_desconto

# 3. Resumo da venda exibido na tela
print("\n--- RESUMO DA VENDA ---")
print("Vendedor responsável:", nome_vendedor)
print("Produto:", nome_produto)
print("Preço original: R$", preco_original)
print("Desconto aplicado:", percentual_desconto, "%")
print("Valor economizado pelo cliente: R$", valor_desconto)
print("Preço final a pagar: R$", preco_final)
```

```python
# 1. Entrada de dados fornecida pelo utilizador
# Usamos int() para transformar a resposta (texto) num número inteiro
quantidade_total_porcoes = int(input("Digite a quantidade total de porções do produto: "))
porcoes_por_dia = int(input("Digite quantas porções consome por dia: "))

# 2. Cálculo da duração em dias
dias_duracao = quantidade_total_porcoes / porcoes_por_dia

# 3. Exibição do resultado na tela
print(f"\nCom consumo de {porcoes_por_dia} porção(ões) por dia, o produto vai durar {dias_duracao:.0f} dias.")
```

```python
# --- SIMULADOR DE COBERTURA DE ESTOQUE (PCP) ---

# 1. Entrada de dados da produção
materia_prima = input("Digite o nome do insumo/matéria-prima: ")
estoque_atual = float(input(f"Digite o estoque atual de {materia_prima} (em un ou kg): "))
consumo_diario = float(input("Digite a demanda/consumo diário estimado: "))

# 2. Cálculo do tempo de cobertura
dias_cobertura = estoque_atual / consumo_diario

# 3. Alerta de suprimentos
print(f"\n--- RELATÓRIO DE COBERTURA DE ESTOQUE ---")
print(f"Item: {materia_prima}")
print(f"Estoque Disponível: {estoque_atual}")
print(f"Consumo Diário: {consumo_diario}")
print(f"Seu estoque atual garante produção para mais {dias_cobertura:.1f} dias.")
```
