```python
class Casa:
    def __init__(self, cor, quartos, banheiros):
        self.cor = cor
        self.quartos = quartos
        self.banheiros = banheiros

    def mostrar_cor(self):
        print(f'A cor da casa é {self.cor}')

    def mostrar_quartos(self):
        print(f'Esta casa tem {self.quartos} quartos')

    def mostrar_banheiros(self):
        print(f'Esta casa tem {self.banheiros} banheiros')

    def adicionar_quarto(self):
        self.quartos += 1
        print(f'Agora esta casa tem {self.quartos} quartos')

    def pintar_casa(self, nova_cor):
        self.cor = nova_cor
        print(f'A cor da casa foi mudada para {self.cor}')

casa1 = Casa('Azul', 4, 2)
casa2 = Casa('Amarelo', 6, 4)

print('\ncasa1:')
casa1.mostrar_cor()
casa1.mostrar_quartos()
casa1.mostrar_banheiros()
casa1.adicionar_quarto()
casa1.pintar_casa('Verde')

print('\nCasa2:')
casa2.mostrar_cor()
casa2.mostrar_quartos()
casa2.mostrar_banheiros()
casa2.adicionar_quarto()
casa2.pintar_casa('Vermelho')
```

```python
class Carro: #Classe que define o objeto

  def __init__(self, cor, ano):#Construtor é o metodo responsável pelos atributos do objeto
    self.cor = cor
    self.ano = ano
    self.ligado = True
    self.seta = None

  def informacoes(self):# Método que define o comportamento do objeto
    print(f'O carro é da cor {self.cor}')
    print(f'O ano do carro é {self.ano}')

  def ligar(self):
    if not self.ligado:
      self.ligado = True
      print('Carro foi ligado')
    else:
      print('Carro já estava ligado')

  def desligar(self):
    if self.ligado:
      self.ligado = False
      print('Carro foi desligado')
    else:
      print('Carro já estava desligado')


  def ligar_seta(self, direcao):# Método que define o comportamento do objeto
    self.seta = direcao
    print(f'Seta ligada para a {self.seta}')

carro1 = Carro('Preto', 2021)
carro1.informacoes()
carro1.desligar()
carro1.ligar_seta('Direita')
```

```python
# Classe que define o objeto
class Pessoa:
    def __init__(self, nome, idade, cargo):  # Construtor é o método responsável pelos atributos do objeto
        self.nome = nome
        self.idade = idade
        self.cargo = cargo

    def informacoes(self):  # Método que define o comportamento do objeto
        print(f'Nome: {self.nome}')
        print(f'Idade: {self.idade}')
        print(f'Cargo: {self.cargo}')

    def promover(self, novo_cargo): # Método que define o comportamento do objeto
        print(f'{self.nome} foi promovido para nova funcao de {novo_cargo}')
        self.cargo = novo_cargo

    def atualizar_idade(self, nova_idade): # Método que define o comportamento do objeto
        if nova_idade > self.idade:
            print(f'Atualizando a idade de {self.idade} para {nova_idade}')
            self.idade = nova_idade
        else:
            print('A nova idade deve ser maior que a anterior')

colaborador1 = Pessoa('Ana', 36, 'Assistente Junior')
colaborador2 = Pessoa('João', 28, 'Dev junior')

colaborador1.informacoes()
colaborador1.promover('Assistente Pleno')
colaborador1.atualizar_idade(37)
colaborador2.informacoes()
colaborador2.atualizar_idade(29)
```

```python
# Classe que define o objeto

class Animal:
  def __init__(self, nome, cor, especie): # Construtor é o método responsável pelos atributos do objeto
    self.nome = nome
    self.cor = cor
    self.especie = especie

  def apresentar(self): # Método que define o comportamento do objeto
    print(f'Eu sou o {self.especie} chamado {self.nome}')

class Gato(Animal):
 def emitir_som(self):
    print('Miau')

class Cachorro(Animal):
  def emitir_som(self):
    print('Au au')

class Elefante(Animal):
  def emitir_som(self):
    print('Toot')

gato1 = Gato('Felix', 'Branco', 'Siames')
gato1.apresentar()
gato1.emitir_som()

cachorro1 = Cachorro('Russo', 'Preto', 'Pastor Alemão')
cachorro1.apresentar()
cachorro1.emitir_som()

Elefante1 = Elefante('Fred', 'Marrom', 'Asiatico')
Elefante1.apresentar()
Elefante1.emitir_som()
```

```python
# Classe que define o objeto

class Pessoa: # Esta é a classe pai
  def __init__(self, nome, idade): # Construtor é o método responsável pelos atributos do objeto
    self.nome = nome
    self.idade = idade

  def apresentar(self): # Método que define o comportamento do objeto
    print(f'Olá, o meu nome é {self.nome} e tenho {self.idade} de idade')

class Funcionário(Pessoa): # Esta é a classe filha (herança)
  def __init__(self, nome, idade, cargo):
    super().__init__(nome, idade) # Chamando o construtor da classe pai
    self.cargo = cargo

  def trabalhar(self):
    print(f'{self.nome} esta trabalhando como {self.cargo}.')



class Cliente(Pessoa):
  def __init__(self, nome, idade, saldo):
    super().__init__(nome, idade) # Chamando o construtor da classe pai
    self.saldo = saldo

  def comprar(self, valor_compra):
    if valor_compra <= self.saldo:
      self.saldo -= valor_compra
      print(f'{self.nome} a sua compra de R${valor_compra} foi aprovada! Saldo atual: R${self.saldo}')
    else:
      print('Saldo insuficiente')


f1 = Funcionário('Maria', 38, 'Gerente de contas')
f1.trabalhar()

c1 = Cliente('Arthur', 26, 200)
c1.apresentar()
c2 = Cliente('Lilian', 32, 300)
c2.comprar(70) # Corrigido: o método correto é 'comprar'
```

```python
# Herança multipla e Herança multivível

# Classe Avo
class Animal: # Construtor é o método responsável pelos atributos do objeto
  def __init__(self, nome):
    self.nome = nome

  def apresentar(self):
    print(f'Eu sou o {self.nome}')

# Classes Pai
class Predador(Animal):
  def cacando(self):
    print(f'O animal {self.nome} esta caçando') # Corrigido f-string

class Presa(Animal): # Corrigido: Agora Presa também herda de Animal
  def fugindo(self):
    print(f'O animal {self.nome} esta fugindo') # Corrigido f-string

# Classes filha
class Coelho(Presa):
  pass

class Tigre(Predador):
  pass

class Golfinho(Predador, Presa):
  pass

coelho1 = Coelho('Teo')
coelho1.fugindo()

tigre1 = Tigre('Leo')
tigre1.cacando()

golfinho1 = Golfinho('Billy')
golfinho1.cacando()
golfinho1.fugindo()
```

```python

# --- CLASSE PAI ---

class Escola():
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
        self.status = True  # Status padrão inicial: Ativo

    def Apresentar(self):
        situacao = 'Ativo' if self.status else 'Inativo'
        print(f'Meu nome é {self.nome} [Status: {situacao}]')

    def desativar(self):
        self.status = False
        print(f'-> O cadastro de {self.nome} foi desativado.')

    def ativar(self):
        self.status = True
        print(f'-> O cadastro de {self.nome} foi ativado.')


# --- CLASSES FILHAS (HERANÇA) ---

class Aluno(Escola):
    def __init__(self, nome, idade, ano):
        super().__init__(nome, idade)
        self.ano = ano

    def Apresentar(self):
        super().Apresentar()
        print(f'Tenho {self.idade} anos, estou no {self.ano}º ano e sou aluno(a).')


class Professor(Escola):
    def __init__(self, nome, idade, materia):
        super().__init__(nome, idade)
        self.titulacao = materia

    def Apresentar(self):
        super().Apresentar()
        print(f'Tenho {self.idade} anos e sou professor(a) de {self.titulacao}.')


class Assistente(Escola):
    def __init__(self, nome, idade, bloco):
        super().__init__(nome, idade)
        self.bloco = bloco

    def Apresentar(self):
        super().Apresentar()
        print(f'Tenho {self.idade} anos e trabalho no {self.bloco}.')


# --- EXECUÇÃO E TESTES ---

print("=== CRIANDO OS OBJETOS ===")
a1 = Aluno('Marcos', 18, 8)
p1 = Professor('Michelle', 39, 'Geografia')
as1 = Assistente('Ana Maria', 29, 'Bloco C')

print("\n=== APRESENTAÇÃO INICIAL ===")
a1.Apresentar()
print("-" * 30)
p1.Apresentar()
print("-" * 30)
as1.Apresentar()

print("\n=== MUDANÇA DE STATUS ===")
p1.desativar()
p1.Apresentar()

print("\n=== REATIVANDO ===")
p1.ativar()
p1.Apresentar()
```
